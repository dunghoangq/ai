# XGBoost - Dense Pipeline

```text
X
↓
QuantileBinner
↓
X_binned
↓
GradientHistogram
```

Or more detailed:

```text
XGBoostClassifier
│
├── QuantileBinner       ← fit once
│
│     ↓
│   X_binned
│
└── Tree 1
     Tree 2
     Tree 3
     ...
        │
        └── DecisionTree
              │
              ├── node
              │    └── GradientHistogram
              │          ↓
              │       best split
              │
              ├── left node
              │    └── new histogram
              │
              └── right node
                   └── new histogram
```
